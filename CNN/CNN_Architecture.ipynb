```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.layers import Dense,Conv2D,Flatten,AveragePooling2D
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_tr,y_tr) ,( X_tst,y_tst) = mnist.load_data()
X_tr = X_tr.reshape(-1, 28, 28, 1)
```

```python
model = Sequential()
```

```python
## LE-NET architecture 

model.add(Conv2D(32,padding = 'valid',kernel_size = (5,5),activation='tanh',input_shape = (28,28,1)))
model.add(AveragePooling2D(pool_size=(2,2),strides = 2,padding = 'valid'))

model.add(Conv2D(16,activation='tanh',kernel_size=(5,5),padding = 'valid'))
model.add(AveragePooling2D(pool_size=(2,2),padding = 'valid',strides =2))

model.add(Flatten())

model.add(Dense(120,activation = 'tanh'))
model.add((Dense(84,activation='tanh')))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics = ['accuracy'],loss = 'sparse_categorical_crossentropy')
```

```python
from sklearn import metrics
model.fit(X_tr,y_tr,validation_split=0.2)
```

```python
pred = model.predict(X_tst)
```

```python
import numpy as np
y_pred = np.argmax(pred,axis = 1)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_tst,y_pred)
```
