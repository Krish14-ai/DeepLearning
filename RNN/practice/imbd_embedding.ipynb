```python
from keras import Sequential
from keras.utils import pad_sequences
from keras.datasets import imdb
from keras.layers import Dense, SimpleRNN, Flatten, Embedding
```

```python
(X_train,y_train), (X_test, y_test) = imdb.load_data(num_words = 10000)
```

```python
print(X_test.shape)
print(X_train.shape)
```

```python
X_train_pad= pad_sequences(X_train, maxlen = 100, padding='post')
X_test_pad = pad_sequences(X_test,padding = 'post', maxlen = 100)
```

```python
model = Sequential()

model.add(Embedding(10000, 2))

model.add(SimpleRNN(32, return_sequences= False))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer= 'adam', loss = 'binary_crossentropy', metrics= ['accuracy'])
```

```python
model.fit(X_train_pad, y_train, validation_split= 0.2, epochs = 5)
```
