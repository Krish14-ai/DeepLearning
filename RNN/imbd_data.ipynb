```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten

from keras.preprocessing.sequence import pad_sequences
```

```python
(X_train,y_train), (X_test, y_test)= imdb.load_data()
```

```python
X_train
```

```python

## all the inputs are different in length
print(len(X_train[1]))
print(len(X_train[2]))
```

```python
## As all the inputs are different we have to perform `Padding`

X_train = pad_sequences(X_train, padding = 'post')
X_test = pad_sequences(X_test, padding= 'post')

print(X_train.shape)
```

```python
model = Sequential()

model.add(SimpleRNN(32, input_shape= (2494,1),return_sequences= False ))
model.add(Dense(1, activation="sigmoid"))
```

```python
model.summary()
```

```python
## Trainig for only one epoch as my Laptop doesnt have a Graphics card :(
model.compile(loss = "binary_crossentropy", optimizer= "adam", metrics= ["accuracy"])
model.fit(X_train, y_train, validation_split= 0.2)
```
