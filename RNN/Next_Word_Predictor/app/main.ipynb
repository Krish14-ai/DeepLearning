```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([docs])
```

```python
tokenizer.word_index
```

```python
input_seq = []

for sentences in docs.split('.'):
     tokenized_sentences= tokenizer.texts_to_sequences([sentences])[0]

     for i in range(1, len(tokenized_sentences)):
         input_seq.append(tokenized_sentences[:i+1])
```

```python
input_seq
```

```python
max_len = max([len(x) for x in input_seq])
```

```python
## Zero padding
padded_inpt_seq = pad_sequences(input_seq, maxlen= max_len, padding = 'pre')
padded_inpt_seq
```

```python
X = padded_inpt_seq[:, :-1]
y = padded_inpt_seq[:,-1]
print(X)
print(y)
```

```python
from keras.utils import to_categorical
y = to_categorical(y,num_classes= 438)
print(y.shape)
```
