```python
import pandas as pd 
from keras.models import Model
```

```python
from keras.layers import *
```

```python
inputA = Input(shape = (32,))
inputB= Input(shape = (128,))
```

```python
## First Branch operates on the first input
X1 = Dense(128, activation = "relu")(inputA)
X2 = Dense(32, activation = "relu")(X1)
```

```python
## second branch operates on the second input 
Y1 = Dense(128,avtivation = "relu")(inputB)
Y2 = Dense(32, activation = "relu")(Y1)
Y3 = Dense(4, activation = "relu")(Y2)
```

```python
## Combine the outputs of two Branches
combined = concatenate(inputs = [X2, Y3])
```

```python
## Apply FC layer and then Regression Prediction on 
## Combined Outputs
Z1 = Dense(2, activation = "relu")(combined)
Z2 = Dense(1, activation = "linear")(Z1)
```

```python
model = Model(inputs = [X2, Y3], outputs  = Z2)
```

```python
from keras.utils import plot_model
plot_model(model,show_shapes  = True)
```
